```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn import svm
from sklearn.metrics import accuracy_score
```

```python
od.download("https://www.kaggle.com/datasets/mathchi/diabetes-data-set")
```

```python
data=pd.read_csv("/content/diabetes-data-set/diabetes.csv")
```

```python
data.head()
```

```python
data.shape
```

```python
data.describe()
```

```python
data['Outcome'].value_counts()
```

0---Nonediabetic
1--->Diabetic

```python
data.groupby('Outcome').mean()
```

```python
x=data.drop(columns='Outcome',axis=1)
y=data['Outcome']
```

```python
print(y)
```

```python
scaler=StandardScaler()
```

```python
scaler.fit(x)
```

```python
standardize_data=scaler.transform(x)
```

```python
print(standardize_data)
```

```python
x=standardize_data
y=data['Outcome']
```

```python
print(x)
print(y)
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,stratify=y,random_state=2)
```

```python
classifir=svm.SVC(kernel='linear')
```

```python
classifir.fit(x_train,y_train)
```

```python
x_train_prediction=classifir.predict(x_train)
```

```python
traning_data_accuracy=accuracy_score(x_train_prediction,y_train)
```

```python
print("Accuracy score of the traning data",traning_data_accuracy)
```

```python
x_test_prediction=classifir.predict(x_test)
testing_data_accuracy=accuracy_score(x_test_prediction,y_test)

```

```python
print("Accuracy score of the traning data",testing_data_accuracy)
```

```python
randome_daata=data.sample(1).iloc[:,:-1]
```

```python
input_data=(randome_daata)
input_data_as_array=np.asarray(input_data)
input_data_reshape = input_data_as_array.reshape(1,-1)
std_data=scaler.fit_transform(input_data_reshape)
```
